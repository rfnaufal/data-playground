### Add new Column (Variables)

```python
reference_date = df['Dt_Customer'].max() #dataset terakhir di 2014
reference_date
```

#### Age, Tenure and No of children

```python
# Create Age
df['Age'] = 2014 - df['Year_Birth'] #2014 krn terakhir dataset di 2014

# Create Customer Tenure in days
df['Customer_Tenure'] = (reference_date - df['Dt_Customer']).dt.days

# Create Children
df['Children'] = df['Kidhome'] + df['Teenhome']

df.describe()
```

#### Total Spending

```python
spending_cols = [
    'MntWines',
    'MntFruits',
    'MntMeatProducts',
    'MntFishProducts',
    'MntSweetProducts',
    'MntGoldProds'
]

df['TotalSpend'] = df[spending_cols].sum(axis=1)
```

#### Total Purchases

```python
purchase_cols = [
    'NumDealsPurchases',
    'NumWebPurchases',
    'NumCatalogPurchases',
    'NumStorePurchases'
]

df['TotalPurchases'] = df[purchase_cols].sum(axis=1)
```

#### Total Accepted Campaign 

```python
campaign_cols = [
    'AcceptedCmp1',
    'AcceptedCmp2',
    'AcceptedCmp3',
    'AcceptedCmp4',
    'AcceptedCmp5',
    'Response'
]

df['CampaignAcceptedTotal'] = df[campaign_cols].sum(axis=1)
```

```python
#Preview
# Preview new variables
df[['Age', 'Customer_Tenure', 'Children', 'TotalSpend', 'TotalPurchases', 'CampaignAcceptedTotal']].head()
```

#### Previous Campaign Accepted Total (all campaign with response)

```python
previous_campaign_cols = [
    'AcceptedCmp1',
    'AcceptedCmp2',
    'AcceptedCmp3',
    'AcceptedCmp4',
    'AcceptedCmp5'
]

df['PreviousCampaignAcceptedTotal'] = df[previous_campaign_cols].sum(axis=1)
df[['ID', 
            'PreviousCampaignAcceptedTotal', 
            'CampaignAcceptedTotal']].head()
```

#### Create Customer Value Segment

```python
# Create customer value segment based on TotalSpend percentile
df['SpendPercentile'] = df['TotalSpend'].rank(pct=True)

def assign_value_segment(p):
    if p >= 0.80:
        return 'High Value'
    elif p >= 0.50:
        return 'Mid Value'
    else:
        return 'Low Value'

df['CustomerValueSegment'] = df['SpendPercentile'].apply(assign_value_segment)

# Check segment distribution
df['CustomerValueSegment'].value_counts(normalize=True) * 100
```

#### Create Retention Segment

```python
# Define thresholds
high_spend_threshold = df['TotalSpend'].quantile(0.80)
medium_spend_threshold = df['TotalSpend'].quantile(0.50)
high_purchase_threshold = df['TotalPurchases'].quantile(0.60)
recency_threshold = 60

def assign_retention_segment(row):
    if row['TotalSpend'] >= high_spend_threshold and row['TotalPurchases'] >= high_purchase_threshold and row['Recency'] < recency_threshold:
        return 'Champions'
    elif row['TotalSpend'] >= high_spend_threshold and row['Recency'] >= recency_threshold:
        return 'At-Risk High Value'
    elif row['TotalSpend'] >= medium_spend_threshold and row['Recency'] < recency_threshold:
        return 'Potential Loyalists'
    elif row['TotalPurchases'] >= high_purchase_threshold:
        return 'Loyal Customers'
    else:
        return 'Low Value'

df['RetentionSegment'] = df.apply(assign_retention_segment, axis=1)

# Check retention segments
df['RetentionSegment'].value_counts()
```

#### Response Comparison

```python
response_comparison = df.groupby('Response').agg(
    CustomerCount=('ID', 'count'),
    AvgIncome=('Income', 'mean'),
    AvgTotalSpend=('TotalSpend', 'mean'),
    AvgTotalPurchases=('TotalPurchases', 'mean'),
    AvgRecency=('Recency', 'mean'),
    AvgChildren=('Children', 'mean'),
    AvgPreviousCampaignAccepted=('PreviousCampaignAcceptedTotal', 'mean')
).round(2)

response_comparison.index = response_comparison.index.map({
    0: 'Non-Responders',
    1: 'Responders'
})

response_comparison
```
